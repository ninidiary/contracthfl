# Figure 7

```python
df_beta1_copy = df_beta1.copy()
df_beta1_copy['sum_of_squares'] = ((1 + df_beta1_copy) ** 2).sum(axis=1) # \sum_n[(1+ \beta_n)^2]
df_beta2_copy = df_beta2.copy()
df_beta2_copy['sum_of_squares'] = ((1 + df_beta2_copy) ** 2).sum(axis=1) # \sum_n[(1+ \beta_n)^2]
df_beta3_copy = df_beta3.copy()
df_beta3_copy['sum_of_squares'] = ((1 + df_beta3_copy) ** 2).sum(axis=1) # \sum_n[(1+ \beta_n)^2]
df_beta4_copy = df_beta4.copy()
df_beta4_copy['sum_of_squares'] = ((1 + df_beta4_copy) ** 2).sum(axis=1) # \sum_n[(1+ \beta_n)^2]
```

```python
beta_square = [
    df_beta3_copy.iloc[1]['sum_of_squares'],
    df_beta4_copy.iloc[1]['sum_of_squares'],
    df_beta1_copy.iloc[1]['sum_of_squares'],
    df_beta2_copy.iloc[1]['sum_of_squares'],
    ]
norm = [
    (np.min(df_acc3['Fw_norm'])) ** 2, 
    (np.min(df_acc4['Fw_norm'])) ** 2,
    (np.min(df_acc1['Fw_norm'])) ** 2,
    (np.min(df_acc2['Fw_norm'])) ** 2
    ]
```

```python
plt.plot(beta_square, norm,'o-', color ='b',linewidth =2 )
plt.xticks(fontsize=14)  
plt.yticks(fontsize=14) 

labels = ['IID', r'$Dir_{8}(1)$', r'$Dir_{8}(0.5)$', r'$Dir_{8}(0.1)$']
offsets = [(0,10), (-30,10),(-55,10), (-70,-10),]
for i, (x, y) in enumerate(zip(beta_square, norm)):
    plt.annotate(labels[i], 
                 xy=(x, y),          # 标注位置（数据坐标）
                 xytext=offsets[i],       # 文本偏移量（像素）
                 textcoords='offset points',  # 偏移量单位
                 fontsize=14,
                 color='k')
    
plt.xlabel(r'$\sum_{n=1}^N (1+\beta_n)^2$',fontsize=14)
plt.ylabel(r'$\min ~ \mathbb{E}[||\nabla F(\omega^{(t,0)})||] $', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('beta_norm.pdf', bbox_inches='tight', transparent=True)
plt.show()
```
